# C1. 
<p>
    Правильнее всего использовать медиану для описания типичной зарплаты во всех группах опыта.
    Дополнительно стоит указывать межквартильный размах или 25-й и 75-й перцентили, чтобы очертить диапазон, в котором лежит большинство предложений.
</p>

## Обоснование
<p> 
    Распределения зарплат во всех группах опыта скошены вправо. Что видно из расчётов в блоке B.
    <b>Из этого следует</b>, 1) Основная масса вакансий сосредоточена на относительно низких зарплата 2) небольшое количество вакансий с очень высокими зарплатами (правый хвост) 3) высокие зарплаты — выбросы или редкие предложения.
</p>
<p>
    Среднее — неудачный выбор метрики для данного распределения, потому что значения хвоста будут сильно искажать метрику, давая неверное представление о зарплатах.
    <b>Медиана</b> делит выборку на 2 равные части, она наилучший выбор, потому что:
    <ul>
        <li>Устойчива к выбросам, т.е не поменяет резко значение при появлении в данных аномалий</li>
        <ul>Даже при сильной асимметрии медиана остаётся репрезентативной (отражает правдивое положение дел в выборке).</ul>
    </ul>
</p>
<p>
    Прочие метрики. Не подходят из-за:
    <ol>
        <li>Мода: может быть не одна, плохо работает с непрерывными данными<li>
        <li>Усечённое среднее: cложнее интерпретировать</li>
        <li>Квартили: не дают одно число для типичной зарплаты</li>
    </ol>
</p>
