<p>Для строгого утверждения проведём формальный статистический тест Манна-Уитни и рассчитаем p-value</p>

```python
# Формируем выборки
group_1_3 = df_clean[df_clean['experience'] == 'От 1 года до 3 лет']['salary']
group_3_6 = df_clean[df_clean['experience'] == 'От 3 до 6 лет']['salary']
```

## Тест Манна-Уитни

```python
import scipy.stats as stats
```

```python
u_stat, p_value_mann = stats.mannwhitneyu(group_1_3, group_3_6, alternative='two-sided')
```

```python
print(f"\nU-статистика: {u_stat:.1f}")
print(f"p-value: {p_value_mann:.6f}")
```

```python
alpha = 0.001
if p_value_mann < alpha:
    print(f"\nРазличие СТАТИСТИЧЕСКИ ЗНАЧИМО на уровне {alpha}")
    print(f"  Зарплаты в группе <От 3 до 6 лет> значимо выше, чем в <От 1 года до 3 лет>")
else:
    print(f"\nазличие НЕ значимо на уровне {alpha}")
```
